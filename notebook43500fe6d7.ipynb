```python
import matplotlib.pyplot as plt
import numpy as np # faster than python lists

# x=[2023, 2024, 2025, 2026]
# y=[15,20,19,40]

x=np.array([2023, 2024, 2025, 2026])
y1=np.array([15,20,19,40])
y2=np.array([17,28,29,5])

line_style=dict(marker="*", ms=20, markerfacecolor="blue", markeredgecolor="red", 
        linestyle="dashdot", linewidth=2)

plt.plot(x,y1, color="green",**line_style) 

plt.plot(x,y2, **line_style) # to customize this, don't repeat from the previous plot, make a dict


plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np # faster than python lists

# x=[2023, 2024, 2025, 2026]
# y=[15,20,19,40]

x=np.array([2023, 2024, 2025, 2026])
y1=np.array([15,20,19,40])
y2=np.array([17,28,29,5])

line_style=dict(marker="*", ms=20, markerfacecolor="blue", markeredgecolor="red", 
        linestyle="dashdot", linewidth=2)

plt.plot(x,y1, color="green",**line_style) 

plt.plot(x,y2, **line_style) # to customize this, don't repeat from the previous plot, make a dict


plt.show()
```

```python
x=np.array([2023, 2024, 2025, 2026])
y1=np.array([15,20,19,40])
y2=np.array([17,28,29,5])
y3=np.array([13,19,102,90])

plt.title("Class Size", fontsize=20,fontweight="bold",family="serif",color="blue")

plt.xlabel("Year", fontsize=20,fontweight="bold",
            family="serif",color="blue" )
plt.ylabel("Students", fontsize=20, family="serif", 
           fontweight="bold", color="blue")

# force ticks
plt.tick_params(axis="both", colors="blue")

plt.plot(x,y1) 

plt.plot(x,y2) # to customize this, don't repeat from the previous plot, make a dict


plt.show()
```

### A bar chart compares distinct categories like types of fruit, countries, or brands. A histogram shows numerical data grouped into continuous ranges or intervals like test scores, heights, or weights

```python
# Bar chart
# np arrays are faster than python lists but list of strings don't benefit from numpy arrays
categories=np.array(["grains", 'fruit', 'vegetables','protein', 'dairy','sweets'])
values=np.array([4,3,2,5,3,1])
#plt.bar(categories, values, color="skyblue")
plt.barh(categories, values, color="skyblue")
                 
```

### Pie Charts

```python
Categories=['Freshmen', 'Sophomores','Juniors','Seniors']
values =np.array([300,250,275,225])
colors=['red', 'yellow','blue','green']

plt.pie(values, labels=Categories, 
       autopct="%1.1f%%", colors=colors, 
       explode=[0,0,0,0.2], 
       shadow=True, 
       startangle=90)

```

### Scatter graphs

```python
x1=np.array([0,1,1,2,3,4,5,6,7,7,8])
y1=np.array([55,60,65,62,68,70,75,78,82,85,87])
x2=np.array([0,1,1,2,2,4,5,6,7,7,0])
y2=np.array([55,60,65,62,68,0,75,78,82,96,97])

plt.scatter(x1,y1, color="red", s=100, label="Class A")
plt.scatter(x2,y2, color="yellow", s=100, label="Class B")

plt.title("Test Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Grade")
plt.legend()

plt.show()
```

### subplots

```python
# Figure = the entire canvas
# Axes =A single plot(subplot), technical an array

x=np.array([1,2,3,4,5])


figure,axes=plt.subplots(2,2)

axes[0,0].plot(x,x*2,color="blue") # can also be .bar
axes[0,0].set_title("x*2")
plt.tight_layout() #Matplotlib function that automatically adjusts subplot parameters so that labels, tick marks, and titles fit neatly within the figure area

```
